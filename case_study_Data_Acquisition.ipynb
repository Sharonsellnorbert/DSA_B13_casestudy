```python
# Explore Population dataset
print("POPULATION")
print("Rows and Columns:", po.shape)
print("Column Names:", po.columns)
print("Data Types:")
print(po.dtypes)
print("Missing Values:")
print(po.isnull().sum())
print("Duplicate Records:", po.duplicated().sum())


# Explore GDP dataset
print("\nGDP")
print("Rows and Columns:", gd.shape)
print("Column Names:", gd.columns)
print("Data Types:")
print(gd.dtypes)
print("Missing Values:")
print(gd.isnull().sum())
print("Duplicate Records:", gd.duplicated().sum())


# Explore Internet Users dataset
print("\nINTERNET USERS")
print("Rows and Columns:", df.shape)
print("Column Names:", df.columns)
print("Data Types:")
print(df.dtypes)
print("Missing Values:")
print(df.isnull().sum())
print("Duplicate Records:", df.duplicated().sum())


# Explore Literacy Rate dataset
print("\nLITERACY RATE")
print("Rows and Columns:", dfx.shape)
print("Column Names:", dfx.columns)
print("Data Types:")
print(dfx.dtypes)
print("Missing Values:")
print(dfx.isnull().sum())
print("Duplicate Records:", dfx.duplicated().sum())
```

```python
po.rename(columns={'country': 'country'})

gd.rename(columns={'Country': 'country'})

df.rename(columns={'country': 'country'})

dfx.rename(columns={'name': 'country'})
```

```python
print(po.dtypes)
print(gd.dtypes)
print(df.dtypes)
print(dfx.dtypes)
```

```python
# Merge Population and GDP
merged = pd.merge(po, gd, on='country')

# Merge with Internet Users
merged = pd.merge(merged, df, on='country')

# Merge with Literacy Rate
merged = pd.merge(merged, dfx, on='country')


merged
```

```python
merged['Internet_penetration']=(merged['internet_users']/merged['population']*100).round(2)
merged['GDP Per Capita']=(merged['GDP']/merged['population']).round(2)
merged
```

```python
import sqlite3

conn = sqlite3.connect('/content/country_statistics.db')

merged.to_sql(
    'country_statistics',
    conn,
    if_exists='replace',
    index=False
)
print("Data successfully loaded into SQLite database.")
```

```python
# Retrieve first 5 records
query = "SELECT * FROM country_statistics LIMIT 5;"

result = pd.read_sql_query(query, conn)

result
```

```python
# 1. Top 5 countries by GDP
print(merged.sort_values('GDP', ascending=False).head(5))
print('---------------------------------------------------------')
# 2. Top 5 countries by population
print(merged.sort_values('population', ascending=False).head(5))
print('---------------------------------------------------------')
# 3. Average literacy rate
print(merged['literacy_rate'].mean())
print('---------------------------------------------------------')
# 4. Literacy rate greater than 90%
print(merged[merged['literacy_rate'] > 90])
print('---------------------------------------------------------')
# 5. Internet penetration greater than 70%
print(merged[merged['Internet_penetration'] > 70])
print('---------------------------------------------------------')
# 6. Top 5 GDP per capita
print(merged.sort_values('GDP Per Capita', ascending=False).head(5))
print('---------------------------------------------------------')
# 7. Total population
print(merged['population'].sum())
print('---------------------------------------------------------')
# 8. Average internet penetration
print(merged['Internet_penetration'].mean())

print('---------------------------------------------------------')
# 9. Country with highest internet users
print(merged.loc[
    merged['internet_users'].idxmax(),
    ['country', 'internet_users']
])
print('---------------------------------------------------------')
# 10. Correlation
print(merged['literacy_rate'].corr(
    merged['Internet_penetration']
))
```
