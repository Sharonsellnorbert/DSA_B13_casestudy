```python
po.rename(columns={'country': 'country'})

gd.rename(columns={'Country': 'country'})

df.rename(columns={'country': 'country'})

dfx.rename(columns={'name': 'country'})
```

```python
print(po.dtypes)
print(gd.dtypes)
print(df.dtypes)
print(dfx.dtypes)
```

```python
# Rename columns so that all datasets have a common key
gd.rename(columns={'Country': 'country'}, inplace=True)
dfx.rename(columns={'name': 'country'}, inplace=True)

# Check column names
print(po.columns)
print(gd.columns)
print(df.columns)
print(dfx.columns)

# Merge Population and GDP
merged = pd.merge(po, gd, on='country')

# Merge with Internet Users
merged = pd.merge(merged, df, on='country')

# Merge with Literacy Rate
merged = pd.merge(merged, dfx, on='country')

# Display merged dataset
display(merged)
```

```python
merged['Internet_penetration']=(merged['internet_users']/merged['population']*100).round(2)
merged['GDP Per Capita']=(merged['GDP']/merged['population']).round(2)
merged
```

```python
import sqlite3

conn = sqlite3.connect('/content/country_statistics.db')

merged.to_sql(
    'country_statistics',
    conn,
    if_exists='replace',
    index=False
)
print("Data successfully loaded into SQLite database.")
```

```python
# Retrieve first 5 records
query = "SELECT * FROM country_statistics LIMIT 5;"

result = pd.read_sql_query(query, conn)

result
```

```python
# 1. Top 5 countries by GDP
print(merged.sort_values('GDP', ascending=False).head(5))
print('---------------------------------------------------------')
# 2. Top 5 countries by population
print(merged.sort_values('population', ascending=False).head(5))
print('---------------------------------------------------------')
# 3. Average literacy rate
print(merged['literacy_rate'].mean())
print('---------------------------------------------------------')
# 4. Literacy rate greater than 90%
print(merged[merged['literacy_rate'] > 90])
print('---------------------------------------------------------')
# 5. Internet penetration greater than 70%
print(merged[merged['Internet_penetration'] > 70])
print('---------------------------------------------------------')
# 6. Top 5 GDP per capita
print(merged.sort_values('GDP Per Capita', ascending=False).head(5))
print('---------------------------------------------------------')
# 7. Total population
print(merged['population'].sum())
print('---------------------------------------------------------')
# 8. Average internet penetration
print(merged['Internet_penetration'].mean())

print('---------------------------------------------------------')
# 9. Country with highest internet users
print(merged.loc[
    merged['internet_users'].idxmax(),
    ['country', 'internet_users']
])
print('---------------------------------------------------------')
# 10. Correlation
print(merged['literacy_rate'].corr(
    merged['Internet_penetration']
))
```

### SQLite Database Analysis (Questions 11 to 16)

```python
import pandas as pd
import sqlite3

# Open database connection
conn = sqlite3.connect('/content/country_statistics.db')

# Ensure the table is created/populated in the DB from the in-memory 'merged' dataframe
merged.to_sql('country_statistics', conn, if_exists='replace', index=False)

# 11. Top 5 countries by GDP
q11 = """
SELECT country, GDP
FROM country_statistics
ORDER BY GDP DESC
LIMIT 5;
"""
print("11. Top 5 countries by GDP:")
display(pd.read_sql_query(q11, conn))
```

```python

q12 = """
SELECT AVG(literacy_rate) AS average_literacy_rate
FROM country_statistics;
"""
print("12. Average literacy rate:")
display(pd.read_sql_query(q12, conn))
```

```python

q13 = """
SELECT country, literacy_rate
FROM country_statistics
WHERE literacy_rate > 90.0;
"""
print("13. Countries with literacy rate above 90%:")
display(pd.read_sql_query(q13, conn))
```

```python

q14 = """
SELECT country, Internet_penetration
FROM country_statistics
ORDER BY Internet_penetration DESC
LIMIT 5;
"""
print("14. Top 5 countries with highest internet penetration:")
display(pd.read_sql_query(q14, conn))
```

```python

q15 = """
SELECT SUM(population) AS total_population
FROM country_statistics;
"""
print("15. Total population:")
display(pd.read_sql_query(q15, conn))
```

```python

q16 = """
SELECT country, "GDP Per Capita"
FROM country_statistics
ORDER BY "GDP Per Capita" DESC
LIMIT 1;
"""
print("16. Country with the highest GDP per capita:")
display(pd.read_sql_query(q16, conn))
```

Summary of the ETL and Analysis Process
Diverse Data Sources:
We extracted data from four distinct file formats: a CSV file for population records (population.csv), an Excel spreadsheet for GDP statistics (gdp.xlsx), a JSON file for internet user counts (internet_users.json), and an XML document for literacy rates (literacy_rate.xml).

Targeted Data Cleaning:
Key column names were standardized (e.g., renaming 'Country' to 'country' and 'name' to 'country') to establish a uniform key for merging, and we validated that there were no duplicate entries or missing values across the tables.

Comprehensive Transformations:
All four datasets were merged on the 'country' column into a unified DataFrame. We then calculated two custom metrics: Internet Penetration (%) (internet users divided by population) and GDP Per Capita (GDP divided by population).

Robust Database Loading:
We established a connection to a local SQLite database and created a table named country_statistics using Python's sqlite3 and pandas libraries, successfully loading all merged data for SQL-based querying. Key Analytical Insights:

GDP & Population:
The United States has the highest total GDP, while China has the largest population in our dataset.

Development Indicators:
Switzerland boasts the highest GDP per capita, and Norway has the highest internet penetration (97.76%) coupled with a 100% literacy rate.

Correlation:
We found a strong positive correlation (~0.77) between a country's literacy rate and its internet penetration percentage.
