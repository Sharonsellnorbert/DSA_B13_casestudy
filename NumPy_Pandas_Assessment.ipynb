```python
# Q11
# Load student_performance.csv into a Pandas DataFrame
student_df = pd.read_csv('/content/student_performance.csv')

print("1. First 5 records:")
display(student_df.head(5))
print('\n_________________________________\n')

print("2. Last 5 records:")
display(student_df.tail(5))
print('\n_________________________________\n')

print("3. Number of rows and columns:")
print(student_df.shape)
print('\n_________________________________\n')

print("4. Column names:")
print(student_df.columns.tolist())
print('\n_________________________________\n')

print("5. Data types:")
print(student_df.dtypes)
print('\n_________________________________\n')

print("6. Statistical summary of numerical columns:")
display(student_df.describe())
```

```python
# Q12

# 1. Filter students with Final_Score > 75
score_gt_75 = student_df[student_df['Final_Score'] > 75]
print("1. Students with Final Score > 75")
display(score_gt_75.head())
print('\n_________________________________\n')

# 2. Filter students with Attendance < 75
attendance_lt_75 = student_df[student_df['Attendance'] < 75]
print("2. Students with Attendance < 75%")
display(attendance_lt_75.head())
print('\n_________________________________\n')

# 3. Filter students in 'Data Science' department with Final_Score > 80
ds_gt_80 = student_df[(student_df['Department'] == 'Data Science') & (student_df['Final_Score'] > 80)]
print("3. Data Science students with Final Score > 80")
display(ds_gt_80.head())
print('\n_________________________________\n')

# 4. Sort the filtered 'Data Science' students by Final_Score in descending order
ds_sorted = student_df.sort_values(by='Final_Score', ascending=False)
print("4. Sorted Data Science students (Final Score ) in descending order:")
display(ds_sorted)
```

```python
# Q13: Feature Creation

# 1.
technical_cols = ['Python_Score', 'SQL_Score', 'Pandas_Score', 'NumPy_Score']
student_df['Average_Technical_Score'] = student_df[technical_cols].mean(axis=1)

# 2. Assign Performance based on Final_Score using a mapping function
def assign_performance(score):
    if score >= 90:
        return 'Outstanding'
    elif score >= 80:
        return 'Excellent'
    elif score >= 70:
        return 'Good'
    elif score >= 50:
        return 'Average'
    else:
        return 'Below Average'

student_df['Performance'] = student_df['Final_Score'].apply(assign_performance)

# Display the updated DataFrame (first 5 records showing newly added columns)
print("DataFrame with newly created 'Average_Technical_Score' and 'Performance' columns:")
display(student_df[['Name', 'Python_Score', 'SQL_Score', 'Pandas_Score', 'NumPy_Score', 'Average_Technical_Score', 'Final_Score', 'Performance']].head())
```

```python
# Q14: GroupBy and Aggregation

# Group by Department and calculate count, mean, min, and max for Final_Score
dept_summary = student_df.groupby('Department')['Final_Score'].agg(['count', 'mean', 'min', 'max'])

# Rename columns for clarity
dept_summary = dept_summary.rename(columns={
    'count': 'Total_Students',
    'mean': 'Average_Final_Score',
    'min': 'Min_Final_Score',
    'max': 'Max_Final_Score'
})

print("Summary of Final Scores by Department:")
display(dept_summary)
```

```python
# Q15
import numpy as np

# Convert column directly to a NumPy array
scores = student_df['Final_Score'].dropna().to_numpy()

# Calculate and print all required statistics
print("--- Final_Score Statistics ---")
print(f"Mean: {np.mean(scores):.2f}")
print(f"Median: {np.median(scores):.2f}")
print(f"Std Dev: {np.std(scores):.2f}")
print(f"25th Percentile: {np.percentile(scores, 25):.2f}")
print(f"75th Percentile: {np.percentile(scores, 75):.2f}")
print("_________________________________\n")

# Filter and display students scoring above the overall mean in one line
above_mean = student_df[student_df['Final_Score'] > np.mean(scores)]
print(f"Students with a Final Score above the overall mean ({np.mean(scores):.2f}) (Count: {len(above_mean)}):")
display(above_mean[['Student_ID', 'Name', 'Department', 'Final_Score']].head())
```

```python
# Q16

student_df = pd.read_csv('/content/student_performance.csv')
student_df['Department'] = student_df['Department'].astype(str).str.strip().str.title()


# 1. Create Average_Score
score_cols = ['Python_Score', 'SQL_Score', 'Pandas_Score', 'NumPy_Score']
student_df['Average_Score'] = student_df[score_cols].mean(axis=1)

# 2. Create Pass_Status
student_df['Pass_Status'] = np.where(student_df['Average_Score'] >= 50, 'Pass', 'Fail')

# Display preview
display(student_df[['Name', 'Average_Score', 'Pass_Status']].head())

# 1. Top 5 students based on Average_Score
top_5 = student_df.sort_values(by='Average_Score', ascending=False).head(5)
print("\n1. Top 5 Students by Average Score:\n")
display(top_5[['Student_ID', 'Name', 'Department', 'Average_Score']])
print('\n_________________________________________\n')
# 2. Department with the highest average score (using Final_Score)
dept_avg_scores = student_df.groupby('Department')['Final_Score'].mean()
best_dept = dept_avg_scores.idxmax()
print(f"\n2. Department with the highest average score: {best_dept} ({dept_avg_scores.max():.2f})")
print('\n_________________________________________\n')
# 3. Calculate average attendance by department
dept_attendance = student_df.groupby('Department')['Attendance'].mean().reset_index()
print("\n3. Average Attendance by Department:")
display(dept_attendance)
print('\n_________________________________________\n')
# 4. Total number of students who passed and failed
pass_fail_counts = student_df['Pass_Status'].value_counts()
print("\n4. Pass/Fail Counts:")
print(pass_fail_counts)

print('\n_________________________________________\n')
# Ensure Pass and Fail columns are represented numerically for aggregation
student_df['Pass_Count'] = (student_df['Pass_Status'] == 'Pass').astype(int)
student_df['Fail_Count'] = (student_df['Pass_Status'] == 'Fail').astype(int)

# Group by Department and aggregate the specified metrics
dept_summary_df = student_df.groupby('Department').agg(
    Total_Students=('Student_ID', 'count'),
    Average_Score=('Average_Score', 'mean'),
    Average_Attendance=('Attendance', 'mean'),
    Pass_Count=('Pass_Count', 'sum'),
    Fail_Count=('Fail_Count', 'sum')
).reset_index()


# Display the final summary DataFrame
print("Summary Dataframe")
display(dept_summary_df)
```

```python
# Q17
import numpy as np
import pandas as pd

# --- SECTION A: Data Preparation ---


# 1. Load the CSV file
sales_df = pd.read_csv('/content/sales.csv')

# 2. Convert Order_Date to datetime handling different formats
sales_df['Order_Date'] = pd.to_datetime(sales_df['Order_Date'], format='mixed', dayfirst=True)

# Clean numeric columns (remove potential currency symbols, commas, or whitespace and convert to float)
numeric_cols = ['Sales', 'Quantity', 'Discount', 'Profit']
for col in numeric_cols:
    if col in sales_df.columns:
        sales_df[col] = sales_df[col].astype(str).str.replace('$', '').str.replace(',', '').str.strip()
        sales_df[col] = pd.to_numeric(sales_df[col], errors='coerce')

# 3. Check and handle missing values
print("Missing values count before cleaning:")
print(sales_df.isnull().sum())

# Fill any missing numerical variables with their mean, and strings with 'Unknown'
for col in numeric_cols:
    sales_df[col] = sales_df[col].fillna(sales_df[col].mean())

string_cols = ['Product', 'Category', 'Region']
for col in string_cols:
    if col in sales_df.columns:
        sales_df[col] = sales_df[col].fillna('Unknown').astype(str).str.strip().str.title()

print("\nData preparation complete. Dimensions of dataset:", sales_df.shape)
display(sales_df.head())


# --- SECTION B: Sales Analysis ---
print("\n___________________________________\n")
# 1. Total sales and profit
total_sales = sales_df['Sales'].sum()
total_profit = sales_df['Profit'].sum()
print(f"1. Total Sales: ${total_sales:,.2f}")
print(f"2. Total Profit: ${total_profit:,.2f}")

# 2. Average sales
avg_sales = sales_df['Sales'].mean()
print(f"3. Average Sales per Order: ${avg_sales:,.2f}")

# 3. Top 5 products based on sales
top_products = sales_df.groupby('Product')['Sales'].sum().reset_index()
top_5_products = top_products.sort_values(by='Sales', ascending=False).head(5)
print("\n4. Top 5 Products by Sales:")
display(top_5_products)

# 4. Region generating the highest sales
region_sales = sales_df.groupby('Region')['Sales'].sum()
best_region = region_sales.idxmax()
print(f"\n5. Region generating the highest sales: {best_region} (${region_sales.max():,.2f})")

```
